The repo is public during development (DEC-009), so no token is needed. If it's made private again, add a Colab secret named `GITHUB_TOKEN` (key icon on the left): a fine-grained token with read-only **Contents** access to `basketball-ai`.


```python
# Get the code (clone the first time, pull after that)
import os, subprocess
from google.colab import userdata

REPO = os.environ.get('BBALL_REPO', 'johnBaloran/basketball-ai')
BRANCH = 'task/0.2-python-env'  # branch under review; becomes 'main' once the task is accepted
WORKDIR = '/content/basketball-ai'
try:
    token = userdata.get('GITHUB_TOKEN')  # only needed if the repo is private
except Exception:
    token = ''
url = f'https://x-access-token:{token}@github.com/{REPO}.git' if token else f'https://github.com/{REPO}.git'
if os.path.isdir(WORKDIR):
    subprocess.run(['git', '-C', WORKDIR, 'fetch', '-q', url, BRANCH], check=True)
    subprocess.run(['git', '-C', WORKDIR, 'checkout', '-q', '-B', BRANCH, 'FETCH_HEAD'], check=True)
else:
    subprocess.run(['git', 'clone', '-q', '--branch', BRANCH, url, WORKDIR], check=True)
    # Don't keep the token in the saved remote URL
    subprocess.run(['git', '-C', WORKDIR, 'remote', 'set-url', 'origin', f'https://github.com/{REPO}.git'], check=True)
del token, url
%cd {WORKDIR}
!git log --oneline -1
```
